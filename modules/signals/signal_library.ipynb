### Momentum

Ranks stocks by their trailing price return over a lookback window. Stocks that have been rising steadily score higher; stocks that have been falling score lower.

The idea: one of the most well-documented anomalies in academic finance; stocks that have performed well over the medium term tend to keep performing well for a while longer before eventually reverting. Thought to stem from investor underreaction to new information, causing prices to adjust gradually rather than instantly.
___
*Definition:*

$$\text{signal} = \frac{\text{Price}_{today}}{\text{Price}_{today - lookback}} - 1$$

No skip period is used (a "12-1" style momentum signal would exclude the most recent month to avoid short-term reversal contamination). This is the simplest possible version, a straight trailing return.
___
*Weaknesses:*
- Highly sensitive to lookback choice. Different windows can produce meaningfully different, even opposite-signed, results
- No sector-awareness. Often, momentum's strength varies significantly by sector
- Prone to sharp reversals during regime shifts (momentum crashes historically follow sudden recoveries after a drawdown)

### Mean Reversion

Ranks stocks by the negative of their recent trailing return. Stocks that fell recently score higher (expected to bounce back); stocks that rose recently score lower.

The idea:short-term price moves are often driven by temporary factors (liquidity imbalances, overreaction to news, forced selling/buying) that tend to partially correct within a short window. This is the classic short-term reversal effect, distinct from momentum, which operates over longer horizons.
___
*Definition:*

$$\text{signal} = -\left(\frac{\text{Price}_{today}}{\text{Price}_{today - lookback}} - 1\right)$$

The negative sign is what flips this from a momentum-style signal into a reversal-style one. This is the exact same trailing-return calculation as momentum, but inverted.
___
*Weaknesses:*
- Highly sensitive to lookback choice, same as momentum. Reversal and momentum can actually flip which one dominates depending on the exact window tested (found directly in prior research: certain medium-term windows behaved like reversal despite being tested as "momentum").
- No sector-awareness
- Short lookback windows can be dominated by noise/microstructure effects rather than actual reversal
